```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("Iris.csv")
```

```python
print(df.head())
```

```python
print(df.describe())
```

```python
print(df.mean(numeric_only=True))
```

```python
print(df.median(numeric_only=True))
```

```python
print(df.std(numeric_only=True))
```

```python
grouped = df.groupby('Species')

print(grouped.mean(numeric_only=True))
```

```python
print(np.percentile(df['SepalLengthCm'], 25))

print(np.percentile(df['SepalLengthCm'], 50))

print(np.percentile(df['SepalLengthCm'], 75))
```

```python
# Variance
print("\nVariance:")
print(df.var(numeric_only=True))
```

```python
# Correlation Matrix
print("\nCorrelation Matrix:")
print(df.corr(numeric_only=True))
```

```python
# Histogram
df.hist(figsize=(10,10))
plt.suptitle("Histograms of Iris Dataset")
plt.show()
```

```python
# Boxplot
plt.figure(figsize=(8,6))
df.boxplot()
plt.title("Boxplot of Iris Dataset")
plt.show()
```

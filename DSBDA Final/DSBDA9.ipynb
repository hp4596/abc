```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = sns.load_dataset('titanic')
```

```python
plt.figure(figsize=(10,7))

sns.boxplot(
    x='sex',
    y='age',
    hue='survived',
    data=df
)

plt.title("Age Distribution by Gender and Survival")

plt.xlabel("Gender")

plt.ylabel("Age")

plt.show()
```

```python
plt.figure(figsize=(8,6))

plt.hist(df['age'].dropna(), bins=20)

plt.title("Age Distribution")

plt.xlabel("Age")

plt.ylabel("Frequency")

plt.show()
```

```python
plt.figure(figsize=(8,6))

sns.kdeplot(
    data=df,
    x='age',
    hue='survived',
    fill=True
)

plt.title("Age Density by Survival")

plt.show()
```

```python
sns.countplot(
    x='sex',
    hue='survived',
    data=df
)

plt.title("Gender vs Survival Count")

plt.show()
```

```python
plt.figure(figsize=(10,7))

sns.violinplot(
    x='sex',
    y='age',
    hue='survived',
    data=df,
    split=True
)

plt.title("Violin Plot of Age Distribution")

plt.show()
```

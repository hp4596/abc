```python
iris = pd.read_csv("iris.csv")
iris.head()
```

```python
iris.columns
```

```python
iris['Species'].unique()
```

```python
species_group = iris.groupby('Species')
```

```python
stats = species_group.describe()
stats
```

```python
for species, data in species_group:
    print("\nSpecies:", species)
    print("Mean:\n", data.mean(numeric_only=True))
    print("Standard Deviation:\n", data.std(numeric_only=True))
    print("Percentiles:\n", data.quantile([0.25, 0.5, 0.75], numeric_only=True))
```
